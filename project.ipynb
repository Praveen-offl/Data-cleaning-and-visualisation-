# Data Cleaning & Visualization Project

To clean, process, and analyze IPL dataset and extract meaningful insights using Python.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("matches.csv")

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
df["date"] = pd.to_datetime(df["date"], dayfirst=True, errors="coerce")
```

```python
df["year"] = df["date"].dt.year
df.head()
```

```python
df["season"].value_counts()
```

```python
df["winner"].value_counts().head(5)
```

```python
df["player_of_match"].value_counts().head(5)
```

```python
(df["toss_winner"] == df["winner"]).value_counts()
```

```python
df["season"].value_counts().sort_index().plot(kind="bar")
plt.title("Matches per Season")
plt.xlabel("Season")
plt.ylabel("Matches")
plt.show()
```

```python
df["winner"].value_counts().head(5).plot(kind="bar")
plt.title("Top 5 Winning Teams")
plt.show()
```

```python
df["player_of_match"].value_counts().head(5).plot(kind="bar")
plt.title("Top Players of the Match")
plt.show()
```


## Conclusion

The dataset was successfully cleaned and analyzed.

Dataset(IPL) analysis is given below:

- Certain teams like Mumbai Indians and Chennai Super Kings dominate wins.
- Matches are evenly distributed across seasons.
